```python
#copy value doesn't take any address
import numpy as np
arr=np.array([1,2,3])

new=arr.copy()
new[1]=100
print("array value is:",arr)
print("copies value is:",new)
```

```python
#view-it access values and address
import numpy as np
arr=np.array([1,2,3])

new=arr.view()
new[1]=100
print("array value is:",arr)
print("viewed value is:",new)
```
