```python
num = int(input("Enter a number value: "))
if num <= 1:
    print("not prime number")
else:
    i = 2
    while i < num:
        if num % i == 0:
            print("not prime number")
            break
        i += 1
    else:
        print("prime number")
```

```python
#variable length
def Student(*marks):
  sum=0
  for i in marks:
    sum=sum+i
    avg=sum/len(marks)

  print("Total num value is:",sum)
  print("Average value is:",avg)
  print("Maximum value is:",max(marks))
  print("Minimum value is:",min(marks))
Student(1,2,3,4,5)
Student(5,10,25,40,55,70,95,110)
```
